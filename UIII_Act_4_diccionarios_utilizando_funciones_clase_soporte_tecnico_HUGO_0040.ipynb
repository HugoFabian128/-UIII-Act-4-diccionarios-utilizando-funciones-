#### Clase soporte tecnico
#### funciones
1. agregar  soporte tecnico en un diccionario
2. Listar diccionarios de soporte tecnico
3. buscar soporte tecnico en diccionario
4. borrar soporte tecnico en diccionario
5. actualizar soporte tecnico en diccionario
6. salir


```python
sistema_tickets = {}

def agregar_ticket():
    id_t = input("ID del ticket: ")
    sistema_tickets[id_t] = {
        "empresa": input("Empresa: "),
        "nivel": input("Nivel: "),
        "tiempo": input("Tiempo: "),
        "software": input("Software: "),
        "dispositivos": input("Dispositivos: ")
    }
    print("Ticket guardado con éxito.")

def Listar_tickets():
    if not sistema_tickets:
        print("\nNo hay tickets registrados.")
        return

    # Encabezado de la tabla con alineación fija
    print("\n" + "="*95)
    print(f"{'ID':<10} | {'EMPRESA':<15} | {'NIVEL':<10} | {'TIEMPO':<10} | {'SOFTWARE':<15} | {'DISPOSITIVOS':<15}")
    print("-" * 95)

    # Filas de datos
    for id_t, info in sistema_tickets.items():
        print(f"{id_t:<10} | {info['empresa']:<15} | {info['nivel']:<10} | {info['tiempo']:<10} | {info['software']:<15} | {info['dispositivos']:<15}")
    print("="*95)

def buscar_ticket():
    id_t = input("ID a buscar: ")
    if id_t in sistema_tickets:
        info = sistema_tickets[id_t]
        print(f"\nDatos del Ticket: {id_t}")
        for clave, valor in info.items():
            print(f"{clave.capitalize()}: {valor}")
    else:
        print("El ID no existe.")

def actualizar_ticket():
    id_t = input("ID del ticket a modificar: ")
    if id_t in sistema_tickets:
        sistema_tickets[id_t]["empresa"] = input("Nueva Empresa: ")
        sistema_tickets[id_t]["nivel"] = input("Nuevo Nivel: ")
        print("Ticket actualizado.")
    else:
        print("El ID no existe.")

def borrar_ticket():
    id_t = input("ID del ticket a borrar: ")
    if id_t in sistema_tickets:
        del sistema_tickets[id_t]
        print("Ticket eliminado.")
    else:
        print("El ID no existe.")

while True:
    print("\n1. Agregar | 2. Listar | 3. Buscar | 4. Actualizar | 5. Borrar | 6. Salir")
    opcion = input("Opción: ")

    if opcion == '1':
        agregar_ticket()
    elif opcion == '2':
        Listar_tickets()
    elif opcion == '3':
        buscar_ticket()
    elif opcion == '4':
        actualizar_ticket()
    elif opcion == '5':
        borrar_ticket()
    elif opcion == '6':
        print("\n" + "*"*45)
        print("Programa realizado por Hugo Fabian NL = 15")
        print("*"*45)
        break
    else:
        print("Opción no válida.")
```
